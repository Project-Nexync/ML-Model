```python
import glob
import os
import pandas as pd
raw_csvs = glob.glob('../data/raw/*.csv', recursive=True)
print("Found CSVs:", raw_csvs)
```

```python
csv_path = raw_csvs[0]
df = pd.read_csv(csv_path)
print("Loaded:", csv_path)
```

```python
df = pd.read_csv(csv_path)

# Standardize column names
df = df.rename(columns={
    'Task Description': 'description',
    'Category': 'category',
    'Skill': 'skills'
})

```

```python
# Exploratory Data Analysis with schema and missing values
df.info()
display(df.describe(include='all').T)
print("Missing values per column:\\n", df.isna().sum())
```

```python
print("Columns:", df.columns.tolist())

if 'task_id' in df.columns: 
    print("Unique tasks:", df['task_id'].nunique())
if 'description' in df.columns: 
    print("Sample description:", df['description'].dropna().sample(3).tolist())
if 'skills' in df.columns:
    print("Example skills cell:", df['skills'].dropna().sample(3).tolist())
```

```python
# Add a unique task_id
df['task_id'] = ['T{:05d}'.format(i) for i in range(len(df))]

# Move task_id to the first column (optional, for readability)
cols = ['task_id'] + [c for c in df.columns if c != 'task_id']
df = df[cols]

# Save updated dataframe to processed folder
df.to_csv('../data/processed/tasks_with_id.csv', index=False)

# Quick check
df.head(5)
```
